```python
# Analisis de modelos - Motor Predictor (notebooks)
print("Preparando entorno...")

import sys
import os
import time

# Asegurar rutas correctas
BASE_DIR = r"C:/Users/delpi/OneDrive/Tesis/ADRpy-VTOL/ADRpy"
ANALISIS_DIR = os.path.join(BASE_DIR, 'analisis')
if ANALISIS_DIR not in sys.path:
    sys.path.insert(0, ANALISIS_DIR)
if os.getcwd() not in sys.path:
    sys.path.append(os.getcwd())

# Importar utilidades (sin reload para rapidez)
print("Importando utilidades...")
start = time.time()
try:
    from Modulos.Analisis_modelos import notebook_utils
    from Modulos.Analisis_modelos.notebook_utils import ControlPanel
    print(f"Importado en {time.time()-start:.2f}s")
except Exception as e:
    print(f"Error importando utilidades: {e}")
    raise SystemExit

# Resolver JSON/LOG con rutas absolutas (mismas reglas que el notebook antiguo)
json_candidates = [
    os.path.join(BASE_DIR, 'analisis', 'Results', 'modelos_completos_por_celda.json'),
    os.path.join(BASE_DIR, 'Results', 'modelos_completos_por_celda.json'),
    os.path.join(os.getcwd(), 'Results', 'modelos_completos_por_celda.json'),
]
json_found = None
for p in json_candidates:
    if os.path.exists(p):
        json_found = p
        break

if json_found:
    notebook_utils.Config.JSON_PATH = json_found
    notebook_utils.Config.LOG_PATH = os.path.join(os.path.dirname(json_found), 'notebook_logs.txt')
    print(f"JSON_PATH => {json_found}")
    print(f"LOG_PATH  => {notebook_utils.Config.LOG_PATH}")
else:
    print("No se encontro 'modelos_completos_por_celda.json' en rutas conocidas:")
    for p in json_candidates:
        print("   -", p)
    print("Ajusta la ruta manualmente mas abajo con el selector.")

print("Iniciando panel de control...")
print("=" * 50)

# Crear y mostrar panel de control interactivo
panel = ControlPanel.create_dashboard()
if panel is not None:
    display(panel)
else:
    print("Panel de texto mostrado arriba")

# Agregar selector opcional de JSON como en el notebook antiguo
try:
    import ipywidgets as w
    from IPython.display import display, clear_output

    lbl = w.HTML("<b>Selecciona el archivo JSON (modelos_completos_por_celda.json):</b>")
    default_json = json_found or getattr(notebook_utils.Config, 'JSON_PATH', '')
    txt = w.Text(value=default_json, description='JSON:', layout=w.Layout(width='80%'))
    btn = w.Button(description='Aplicar', button_style='success', tooltip='Actualizar rutas en notebook_utils.Config')
    info = w.Output()

    @btn.on_click
    def _apply(_):
        path = txt.value.strip().strip('"')
        with info:
            clear_output(wait=True)
            if not os.path.exists(path):
                print(f"Archivo no existe: {path}")
                return
            notebook_utils.Config.JSON_PATH = path
            notebook_utils.Config.LOG_PATH = os.path.join(os.path.dirname(path), 'notebook_logs.txt')
            print(f"JSON_PATH: {notebook_utils.Config.JSON_PATH}")
            print(f"LOG_PATH : {notebook_utils.Config.LOG_PATH}")
            print("Si el panel ya estaba abierto, relanzalo para usar la nueva ruta.")

    ui = w.VBox([lbl, txt, btn, info])
    display(ui)
except Exception as e:
    print(f"No se pudo crear el selector de JSON: {e}")

print("\nResumen del panel:")
print("  - Lanzamiento con apertura automatica del navegador")
print("  - Diagnostico de callbacks e IDs") 
print("  - Debug de clicks en tiempo real")
print("  - Gestion de logs con confirmacion")
print("  - Estado del sistema unificado")
print("  - Deteccion de IDs problematicos mejorada")
```
