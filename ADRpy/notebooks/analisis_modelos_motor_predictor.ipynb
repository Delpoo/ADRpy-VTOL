## 🧭 ¿Qué es esta herramienta?
Una app web integrada al notebook para **explorar modelos** con:

## 🧩 Elegí la **celda** a estudiar (concepto Excel)
La **celda** es la combinación: **Aeronave** + **Parámetro**.  
Ej.: `AeronaveX | MTOW` → foco de análisis sobre ese par.

---

## 🎛️ Panel izquierdo — Filtros y controles (qué hace cada uno)

- 🛩️ **Aeronave** (dropdown)  
  Selecciona el conjunto de datos (la aeronave) a estudiar. Actualiza las demás opciones.

- 📏 **Parámetro** (dropdown)  
  Variable objetivo para esa aeronave (p. ej., *MTOW*, *alcance*, etc.).  
  > **Aeronave + Parámetro = Celda activa**.

- 🔎 **Predictor** (dropdown)  
  Muestra **solo** modelos que usan ese predictor.  
  → Elegí **“Todos los predictores”** para verlos sin restricción.

- 🧪 **Tipos de modelo** (checklist)  
  Activa las familias a comparar (ej.: `linear-1`, `poly-1`, `linear-2`, `poly-2`).  
  → Útil para **comparar enfoques** sobre la misma celda.

- 🧩 **Métodos de imputación** (checklist)  
  Decide qué capas ver: `final`, `similitud`, `correlacion`.

- ⚖️ **Tipo de comparación** (radio)  
  Cambia el **criterio global** de comparación (por tipo, top globales, # de predictores, etc.).

- 🎚️ **Opciones de visualización** (toggles)
  - **Puntos de entrenamiento** ↔ datos reales usados para ajustar.  
  - **Puntos teóricos** ↔ evaluados desde la ecuación del modelo.  
  - **Puntos de imputación** ↔ marcadores especiales (final/similitud/correlación).  
  - **Curvas del modelo** ↔ líneas ajustadas visibles/ocultas.  
  - **Solo tramo con datos reales** ↔ evita extrapolaciones “vacías”.  
  - **Incluir sin LOOCV** ↔ muestra modelos sin esa validación (pueden ser menos confiables).  
  - **Ocultar leyenda** ↔ más espacio de gráfico.

> 📱 En pantallas chicas, podés **plegar/desplegar** el panel de filtros.

---

## 📌 Pestañas (barra superior)

- **2D** 📈  
  Curva(s) por modelo + puntos de entrenamiento/teóricos/imputación.  
  → Usá el **hover** para ver valores y ecuaciones resumidas.

- **3D** 📊  
  Superficie del modelo **solo si hay 2 predictores** (lineal o polinómico).  
  → Rotá con el mouse, acercá/alejá y explorá el relieve.

- **Métricas/Resumen** 📋  
  Indicadores globales: modelos válidos, cobertura por tipo, faltantes, etc.

---

## 🖼️ Zonas de la interfaz

- **Lienzo central (gráfico)**  
  - Zoom **persistente** entre cambios.  
  - **Leyenda** clicable (mostrar/ocultar series).  
  - **Exportar PNG** con el botón de cámara.

- **Panel derecho (información)**  
  Estado de la celda, notas y **advertencias** (qué falta o qué limitar).

- **Tabla inferior (resumen)**  
  Datos y resúmenes relacionados con lo que estás viendo.  
  - Resalta **❌ críticos** y **⚠️ warnings**.  
  - Podés **seleccionar filas** para enfocar un modelo.

---

## ✅ Buenas prácticas de lectura

- **Tabla vacía** → cambiá Aeronave/Parámetro o revisá filtros activos.

## 🧱 Capas del sistema (qué hace cada una)

### 1) Lanzamiento (Entrar a la app)
- **launch_app.py**  
  - Define rutas, puertos y modo (*debug/producción*).  
  - Llama a la app principal para iniciar la interfaz.

### 2) Datos (Leer, unificar, estructurar)
- **data_loader.py**  
  - Abre el JSON y repara valores no estándar (NaN/Infinity).  
  - Acepta **estructura nueva** y **antigua**, entregando un formato unificado.
- **json_data_helpers.py**  
  - Expone **modelos_por_celda** y **detalles_por_celda** listos para usar.  
  - Si falta un DataFrame original, **reconstruye** uno a partir de entrenamiento (columnas X, valores X, y_original).

### 3) Análisis y Validación (Confiar antes de ver)
- **equation_analyzer.py**  
  - Identifica tipo de ecuación y número de predictores (¿es graficable en 2D/3D?).
- **equation_validator.py**  
  - Reglas clave: normalización por modelo, compatibilidad de dimensiones, consistencia ecuación–datos, “visibilidad” de términos polinómicos.  
  - Resultado: **errores críticos** (bloquean gráfico) y **warnings** (se muestran en UI).

### 4) Visualización y UI (Entender de un vistazo)
- **plot_interactive.py** (2D)  
  - Curvas de modelos, puntos de entrenamiento/teóricos/imputación, *hover* informativo.
- **plot_3d.py** (3D)  
  - Superficies cuando hay **dos predictores**; rangos seguros para X1/X2/Y.
- **metrics_dashboard.py**  
  - KPIs y tablas de estado/cobertura/faltantes.
- **ui_components.py**  
  - Filtros (izquierda), panel de información (derecha), tabla inferior, modales.
- **plot_config.py**  
  - Colores, símbolos y estilo por familia de modelos.
- **plot_stability.py**  
  - Parámetros de interacción con **estado persistente** de zoom/leyenda (*uirevision*).
- **normalization_engine.py**  
  - Mapeo **normalizado ↔ original** usando datos de entrenamiento (nunca escalas globales).
- **utils.py**  
  - Rangos “seguros” para graficar, sanitización, validaciones numéricas y mensajes estándar.
- **main_visualizacion_modelos.py**  
  - Arma la **página**, registra **callbacks** y orquesta todo lo anterior.

---

## 🔁 Qué pasa cuando cambiás un filtro (viaje del dato)

1. Elegís **Aeronave/Parámetro** (definís la **celda**).  
2. La app consulta **modelos_por_celda / detalles_por_celda** para esa celda.  
3. Se reconstruye/actualiza el **DataFrame** si es necesario (entrenamiento → DF).  
4. Se **validan** los modelos (críticos/warnings).  
5. Si hay críticos → se muestra el **motivo** y no se dibuja.  
6. Si todo ok → se **rinden** las vistas:  
   - **2D** (curvas + puntos) y/o **3D** (superficie con 2 predictores).  
7. El **panel derecho** muestra contexto/alertas y la **tabla inferior** resume datos/modelos.  
8. El **zoom**/estado se conserva para no “perder” la inspección actual.

---

## 🧭 Principios de diseño (por qué así)

- <span style="background:#E8F5E9;color:#1B5E20;padding:0 6px;border-radius:8px;font-weight:600">Escala original</span>  
  Interpretación directa; la normalización vive solo “puertas adentro” para estabilidad numérica.
- <span style="background:#E3F2FD;color:#0D47A1;padding:0 6px;border-radius:8px;font-weight:600">Robustez de lectura</span>  
  Soporta JSON mixtos, reconstruye datos y explica qué falta si no puede graficar.
- <span style="background:#FFF3E0;color:#E65100;padding:0 6px;border-radius:8px;font-weight:600">UX estable</span>  
  **`uirevision`** mantiene zoom/leyenda; cambios de filtros no “rompen” tu vista.
- <span style="background:#FCE4EC;color:#880E4F;padding:0 6px;border-radius:8px;font-weight:600">Separación de responsabilidades</span>  
  Carga ≠ Validación ≠ Visualización ≠ UI. Facilita mantenimiento y pruebas.

---

## 🧪 Taxonomía de mensajes

- **❌ Crítico (bloquea):**  
  Faltan columnas/coeficientes, dimensiones incompatibles, evaluaciones inválidas (NaN/Inf).
- **⚠️ Warning (sigue pero avisa):**  
  Rangos pobres, coeficientes con escalas extremas, baja cobertura, términos polinómicos “invisibles”.

Los verás en el **panel derecho** y/o en la **tabla inferior**, marcados para esa celda/modelo.

---

## 🔌 Extensiones habituales (dónde tocar)

- **Nuevas familias de modelo:** sumar estilo en `plot_config.py` y lógica en **plot_interactive/plot_3d**.  
- **Métricas adicionales:** añadir al **metrics_dashboard** y alimentar desde validadores.  
- **Vistas personalizadas (>2 predictores):** cortes 2D/heatmaps, habilitados por tipo de modelo.  
- **Enriquecer datos:** más metadatos en **detalles_por_celda** (unidades, notas, límites físicos).

---

## 🧠 Glosario mínimo

- **Celda:** combinación **Aeronave | Parámetro** (tu foco de análisis).  
- **Predictor:** variable independiente de un modelo (X, X1, X2).  
- **Imputación:** valor estimado cuando falta un dato (final/similitud/correlación).  
- **Cobertura:** cuántos modelos/datos válidos hay para un conjunto dado.
