## 🧭 ¿Qué es esta herramienta?
Una interfaz interactiva (panel lateral + área de resultados) que te permite:
- 🧮 **Similitud configurable**: pesos por parámetro, modos (Ignorar/Mín/Máx/Fijo), penalización de NaN y **α** (exigencia de distancia).
- 🏷️ **Segmentación opcional**: Global / Filtrar por valor / Preferir valor (con factor de preferencia).
- 💡 **Sugerencias Top-K**: valores recomendados por parámetro a partir de vecinos cercanos (con controles IQR y ponderación por 1/dist).
- 📈 **Tendencias X-Y**: dispersión, líneas guía y filtros por IQR y mínimo de muestras.
- 🚨 **Outliers**: resumen global y exploración por columna con LOW/HIGH basados en IQR.
- 🧾 **Narrativa/Informe**: síntesis editable/exportable (Markdown/HTML) de tu sesión.

👉 Tu trabajo es **elegir bien los parámetros y sus reglas**; la herramienta orquesta el resto.

## ⚡ Cómo iniciar (10 segundos)
1. Ejecutá la celda que **lanza la app** (indicada en el notebook).  
2. Se abrirá la UI con **panel izquierdo** (controles) y **zona central** (resultados).  
3. Activá algunos **parámetros clave**, definí **pesos y modos**, ajustá **α** y **Top-K**, y presioná **▶ Recalcular**.

> 💡 Consejos rápidos:  
> • Empezá con 3–6 parámetros **muy informativos**.  
> • Si el Top-K queda chico, **bajá α**, relajá restricciones (pasá de Fijo → Mín/Máx) o desactiva parámetros con muchos NaN.

---

## 🧩 ¿Cómo elijo la “celda” de interés?
La lógica es similar a “Aeronave × Parámetro” (o el par que represente tu foco).  
- **Seleccioná tu objetivo** (la fila/ítem que querés comparar) en los controles iniciales.  
- Ajustá **parámetros, modos y pesos** que definan la similitud respecto a ese objetivo.

---

## 🎛️ Panel izquierdo — Filtros y controles (qué hace cada uno)

- 🧮 **α (alpha) de similitud**  
  Controla cuán estricta es la distancia. **Más alto = más exigente** (menos vecinos relevantes).
  
- 🧲 **Top-K**  
  Cantidad de **vecinos** que se usan para sugerencias y análisis.

- 🧱 **Penalizar NaN** + **Penalidad**  
  Si un vecino tiene faltantes en **parámetros activos**, se le **suma castigo** en la distancia (baja su ranking).

- ✂️ **Quitar atípicos (IQR)** + **factor IQR**  
  Excluye del Top-K valores fuera de `[Q1 − f·IQR, Q3 + f·IQR]`. Ayuda a estabilizar.

- 🧭 **Segmentación (opcional)**  
  **Segementar por** una columna categórica y elegí **Modo**:  
  • **Global**: no segmenta.  
  • **Filtrar por valor**: usa solo ese segmento.  
  • **Preferir valor**: prioriza ese segmento pero permite otros (ajustá **factor prefer**).

- ▶ **Acciones**  
  **Recalcular** (aplica cambios), **🧹 Limpiar** (restablece), **Auto** (recalcula al mover algo).

---

## 🧩 Panel dinámico de parámetros (izquierda)

Cada parámetro tiene su propia fila con:
- **✅ Activar**: incluir/excluir en la similitud.  
- **Modo**: `Ignorar · Mínimo · Máximo · Fijo`.  
  - En `Mín/Máx/Fijo` aparece un **campo de valor** (cota/objetivo).  
- **Peso**: da **más/menos importancia** relativa.  
- **ℹ️ Info**: abre una ficha con **distribución**, **IQR**, **outliers** y pequeñas notas de contexto.

Herramientas del encabezado:
- **Buscar parámetro…**, **Mostrar solo activos**, **Activar visibles**, **Desactivar todos**.

> 💡 Buen uso de modos:  
> • `Fijo` es potente pero **estricto** (podés quedarte sin vecinos).  
> • Preferí `Mín/Máx` cuando quieras orientar sin bloquear.

---

## 📌 Pestañas / Acordeones (barra superior o central)

### 1) **Ranking** 🏅
Lista ordenada de **vecinos más similares** a tu objetivo teniendo en cuenta **α, pesos, modos, NaN y segmentación**.  
**Lectura:** arriba = mejor. Si ves muy pocos vecinos, **relajá** restricciones o bajá α.

### 2) **Sugerencias** 💡
Tres vistas complementarias:  
- **Resumen** por parámetro: valor sugerido (promedio ponderado/mediana), IQR y notas (“poca diversidad”, “excluidos por IQR”).  
- **Detalle**: elegís un parámetro y ves **histograma/box** del Top-K con Q1/Q3/IQR y la **posición del objetivo**.  
- **Vecinos**: tabla del Top-K con distancias y flags (outlier/segmento).  
Controles propios: **Top-K local**, **quitar atípicos**, **factor IQR**, **ponderar 1/dist**, **Recalcular**.

### 3) **Tendencias X-Y** 📈
Elegís **X** e **Y** (numéricos) para ver **dispersión** y (opcional) línea guía.  
Filtros: **IQR** y **mínimo de muestras**. (Opcional) color por **segmento** o **pertenencia al Top-K**.

### 4) **Outliers (IQR)** 🚨
- **Resumen global**: ordena columnas por **cantidad de atípicos**.  
- **Explorador**: histograma por columna con **LOW/HIGH** (IQR) + lista de outliers.  
Ajustes: **factor IQR**, **mínimo de muestras**, **n° de bins**.

### 5) **Narrativa / Informe** 🧾
Genera un **resumen estructurado**:  
contexto (segmentación, α, Top-K, IQR, parámetros activos), **hallazgos** (sugerencias, outliers, tendencias) y **recomendaciones**.  
Permite **exportar** a Markdown/HTML.

---

## ✅ Buenas prácticas de lectura
- **Primero densidad**: asegurate de tener suficiente muestra antes de concluir.  
- **IQR al rescate**: si ves picos raros, activa **quitar atípicos** y ajustá el **factor**.  
- **Pesos con criterio**: subí el peso solo en parámetros con **fuerte señal**.  
- **Segmentación “Preferir”** suele ser más flexible que “Filtrar por valor”.  
- **No confundas correlación con causalidad**: las tendencias dan contexto, no pruebas causales.

4. Presioná **▶ Recalcular** (o activá **Auto**).  
5. Mirá el **Ranking** → ¿hay suficientes vecinos y pocos NaN penalizados?

8. Explorá **Tendencias X-Y** para validar que lo sugerido es **razonable** en tu dominio.  
9. Generá la **Narrativa** (exportable) para documentar y compartir decisiones.

---

## 🆘 Problemas comunes y soluciones
- **Top-K demasiado chico / sin vecinos** → bajá **α**, desactiva algunos parámetros o cambiá `Fijo` por `Mín/Máx`.  
- **Muchos NaN penalizados** → reducí **peso** de esos parámetros o desactivalos.  
- **Sugerencias inestables** → activá **quitar atípicos** y ajustá **factor IQR**; probá **ponderar 1/dist**.  
- **Tendencias sin puntos** → elegí columnas numéricas y subí el **mínimo de muestras**.

## 🧠 Glosario mínimo

- **α**: exigencia de similitud (↑ = más estricta).  
- **IQR**: rango intercuartílico; base para detectar outliers.  
- **Top-K**: cantidad de vecinos usados para sugerir.  
- **Preferir valor**: segmentación “blanda”; prioriza pero no excluye.

## 🧱 Capas del sistema (qué hace cada una)

### 1) Lanzamiento (Entrar a la app)
- **main.py**  
  - Dibuja la UI (panel izquierdo + área central + acordeones).  
  - Lee **controles globales** (α, Top-K, IQR, segmentación) y el **panel de parámetros** (activar, modo, valor, peso).  
  - **Orquesta** el pipeline y actualiza resultados (Ranking, Sugerencias, Tendencias, Outliers, Narrativa).
- **guias.py**  
  - Textos de onboarding y **tooltips unificados** para cada control (explican qué hace y cuándo usarlo).

### 2) Datos y análisis (Calcular con criterio)
- **similitud.py**  
  - Calcula la **distancia ponderada** aplicando **modos** (Ignorar/Mín/Máx/Fijo) como restricciones **suaves**.  
  - Maneja **penalización de NaN** y **segmentación** (Global / Filtrar / Preferir + factor).  
  - Devuelve **Ranking** y **Top-K** (con opción de excluir atípicos por IQR).
- **sugerencias.py**  
  - Sobre el **Top-K**, computa **valores sugeridos** por parámetro (media/mediana, **ponderar 1/dist**, recortes por IQR).  
  - Expone **resumen** por parámetro y **detalle** con vecinos usados.
- **tendencias.py**  
  - Vistas **X–Y** (dispersión, línea guía), con filtros por **IQR** y **mínimo de muestras** para estabilidad.  
  - (Opcional) color por **segmento** o **pertenencia al Top-K**.
- **outliers.py**  
  - Detecta **atípicos** por IQR (global y por columna), genera **LOW/HIGH** y listados.  
- **verificacion.py**  
  - Chequeos de **consistencia** (columnas, cardinalidad, %NaN), alertas si el set está **sobre-restringido** o **ruidoso**.

### 3) Presentación y narrativa (Contar la historia)
- **mplutils.py**  
  - Asegura **consistencia visual** (escalas, etiquetas, layouts).  
- **narrativa.py**  
  - Arma un **informe** con **contexto** (α, Top-K, IQR, segmentación, parámetros activos), **hallazgos** (sugerencias, outliers, tendencias) y **recomendaciones**.  
  - Permite **exportar** a Markdown/HTML.

---

## 🔁 Flujo del programa (paso a paso)

1. El usuario define **α, Top-K, IQR, segmentación** y **modos/pesos** por parámetro.  
2. **main.py** recoge la configuración y lanza el cálculo.  
3. **similitud.py** aplica segmentación, restricciones por parámetro y **penalización de NaN**.  
4. Se calcula la **distancia ponderada** y se arma el **Ranking**; se deriva el **Top-K** (con IQR si está activo).  
5. Con el Top-K, **sugerencias.py** calcula **valores sugeridos** por parámetro (ponderaciones, recortes).  
6. **outliers.py** genera el **resumen global** y **detalle por columna** con LOW/HIGH.  
7. **tendencias.py** construye **X–Y** con filtros (IQR, mínimo de muestras).  
8. **main.py** renderiza tablas y gráficos; **narrativa.py** sintetiza y **exporta** el informe.

---

## 🧬 Datos intermedios (qué se guarda y para qué)

- **Config global**: {α, Top-K, IQR_on, IQR_factor, penalizar_NaN, segmentar_por, modo_segmento, factor_prefer}  
- **Config por parámetro**: {activo, modo, valor (si aplica), peso}  
- **Ranking**: distancia, penalizaciones, flags (segmento/outlier)  
- **Top-K**: subconjunto filtrado (IQR/min_muestras)  
- **Sugerencias**: tabla por parámetro (sugerido, IQR, estabilidad, notas)  
- **Outliers**: resumen (columna → #atípicos) + detalle (LOW/HIGH, lista)  
- **Tendencias**: X, Y, filtros y estilos de color  
- **Narrativa**: bloques de contexto + hallazgos + próximos pasos
